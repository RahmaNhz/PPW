```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/PPW/tugas/preprocessing-cnnnews.csv")
df.head()
```

## Split Data

```python
import pandas as pd
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['stopword_removal'], df['kategori'], test_size=0.2, random_state=42)
```

## TFIDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)  # Fit-transform pada data latih
X_test_tfidf = vectorizer.transform(X_test)        # Transformasi data uji dengan model TF-IDF yang sama

# Mengubah hasilnya menjadi DataFrame
X_train_tfidf_df = pd.DataFrame(X_train_tfidf.toarray(), columns=vectorizer.get_feature_names_out())
print("TFIDF: DATA TRAINING")
X_train_tfidf_df
```

```python
# Mengubah hasil transformasi X_test_tfidf menjadi DataFrame
X_test_tfidf_df = pd.DataFrame(X_test_tfidf.toarray(), columns=vectorizer.get_feature_names_out())
print("TFIDF: DATA TESTING")
X_test_tfidf_df
```

## SVD

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

svd = TruncatedSVD(n_components=50, random_state=42) # komponen
X_train_svd = svd.fit_transform(X_train_tfidf)
X_test_svd = svd.transform(X_test_tfidf)

print(X_test_svd.shape)
```

```python
X_train_svd_df = svd_df = pd.DataFrame(X_train_svd)
X_train_svd_df
```

## Penerapan Logistic Regression

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
```

```python
# kolom yang berisi label target
y = df['kategori']
```

```python
# Membuat model Logistic Regression
logreg = LogisticRegression()
logreg.fit(X_train_svd, y_train)
```

```python
y_pred = logreg.predict(X_test_svd)
```

```python
# Hitung metrik evaluasi
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')  # Gunakan average='weighted' untuk kelas tidak seimbang
recall = recall_score(y_test, y_pred, average='weighted')

# Tampilkan hasil
print(f"Akurasi: {accuracy:.4f}")
print(f"Presisi: {precision:.4f}")
print(f"Recall: {recall:.4f}")

# Atau tampilkan laporan klasifikasi lengkap
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
